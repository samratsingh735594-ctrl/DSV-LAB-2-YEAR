```python
import pandas as pd
students_df = pd.read_csv("students_master.csv")
marks_df = pd.read_csv("exam_marks.csv")


print(" students_master.csv (First 5 rows) ")
display(students_df.head())

print("\n exam_marks.csv (First 5 rows)")
display(marks_df.head())

common_columns = list(
    set(students_df.columns).intersection(set(marks_df.columns))
)
print(f"\nCommon column(s): {common_columns}")
```

```python

inner_merged_df = pd.merge(students_df, marks_df, on="Student_ID", how="inner")

print("Inner Merge Result")
display(inner_merged_df)
```

```python
left_merged_df = pd.merge(students_df, marks_df, on="Student_ID", how="left")

print("Left Merge Result ")
display(left_merged_df)
```

```python
right_merged_df = pd.merge(students_df, marks_df, on="Student_ID", how="right")


print("Right Merge Result")
display(right_merged_df)
```

```python
outer_merged_df = pd.merge(students_df, marks_df, on="Student_ID", how="outer")

print(" All Unique Student_IDs (Outer Merge)")
display(outer_merged_df[["Student_ID"]])
```

```python
ca_df = pd.read_csv("section_CA.csv")
cb_df = pd.read_csv("section_CB.csv")

concatenated_df = pd.concat([ca_df, cb_df], axis=0)
print("Vertically Concatenated DataFrame (Original Indexes Preserved)")
display(concatenated_df)

print("\n Row Index Values")
print(concatenated_df.index.tolist())
```

```python
concatenated_df = pd.concat([ca_df, cb_df], axis=0, ignore_index=True)
print("Vertically Concatenated DataFrame (ignore_index=True)")
display(concatenated_df)

print("\nNew Index ")
print(concatenated_df.index)
```

```python
concatenated_axis1 = pd.concat([ca_df, cb_df], axis=1)
print("Horizontally Concatenated DataFrame (axis=1)")
display(concatenated_axis1)

print("\n Column Labels")
print(concatenated_axis1.columns.tolist())
```

```python
attendance_series = ca_df["Attendance"]
student_id_series = cb_df["Student_ID"]
concatenated_series = pd.concat([attendance_series, student_id_series], axis=0)
print("Concatenated Series")
display(concatenated_series)
```

```python

original_df = pd.read_csv("marks_original.csv")
updated_df = pd.read_csv("marks_updated.csv")
original_indexed = original_df.set_index("Student_ID")
updated_indexed = updated_df.set_index("Student_ID")


combined_df = original_indexed.combine_first(updated_indexed)


print("--- Combined Marks (using combine_first with Index) ---")
display(combined_df)
```

```python
import numpy as np


original_df = pd.read_csv("marks_original.csv").set_index("Student_ID")
updated_df = pd.read_csv("marks_updated.csv").set_index("Student_ID")

combined_max_df = original_df.combine(updated_df, np.maximum)

print("--- Combined Marks using np.maximum ---")
display(combined_max_df)
```

```python
original_df = pd.read_csv("marks_original.csv")
updated_df = pd.read_csv("marks_updated.csv")

direct_combine = original_df.combine_first(updated_df)

print("--- Direct combine_first() (Without Index Alignment) ---")
display(direct_combine.head())

original_indexed = pd.read_csv("marks_original.csv").set_index("Student_ID")
updated_indexed = pd.read_csv("marks_updated.csv").set_index("Student_ID")

indexed_combine = original_indexed.combine_first(updated_indexed)

print("\n--- Indexed combine_first() (With Student_ID Alignment) ---")
display(indexed_combine.head())
```

```python
original_df = pd.read_csv("marks_original.csv").set_index("Student_ID")
updated_df = pd.read_csv("marks_updated.csv").set_index("Student_ID")

combined_max_marks = original_df.combine(updated_df, np.fmax)

sorted_combined_marks = combined_max_marks.sort_index(ascending=True)

print("--- Final Combined Higher Marks (Ascending Student_ID Order) ---")
display(sorted_combined_marks)
```

```python
long_df = pd.read_csv("marks_long.csv")
wide_marks_df = long_df.pivot(
    index="Student_ID", columns="Subject", values="Marks")
stacked_df = wide_marks_df.stack()

reconstructed_wide_df = stacked_df.unstack()

is_identical = wide_marks_df.equals(reconstructed_wide_df)

print("--- Reconstructed Wide Table ---")
display(reconstructed_wide_df)

print(f"\nDoes the reconstructed table match the original pivot table? {is_identical}")
```
