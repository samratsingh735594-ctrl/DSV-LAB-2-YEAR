Q26


```python
print("Before replacement:")
print(df[["branch", "product", "customer_rating"]])

df["customer_rating"] = df["customer_rating"].fillna(
    df.groupby(["branch", "product"])["customer_rating"].transform("mean")
)

print("\nAfter replacement:")
print(df[["branch", "product", "customer_rating"]])
```

Q27

```python
def weighted_avg_discount(group):
    total_units = group["units_sold"].sum()
    if total_units == 0:
        return 0
    return (group["units_sold"] * group["discount_pct"]).sum() / total_units

df.groupby("branch").apply(weighted_avg_discount)
```

Q28

```python
top_3_records = df.groupby("branch", group_keys=False).apply(
    lambda x: x.nlargest(3, "units_sold")
)
top_3_records[["branch", "product", "day", "units_sold"]]
```

Q29

```python
df["sales_contribution_pct"] = (
    df["units_sold"] / df.groupby("branch")["units_sold"].transform("sum")
) * 100

print(df.groupby("branch")["sales_contribution_pct"].sum())
```

Q30

```python
df["sales_category"] = pd.qcut(
    df["units_sold"],
    q=4,
    labels=["Low Sales", "Below Average Sales", "Above Average Sales", "High Sales"]
)

df.groupby("sales_category", observed=False).agg(
    total_units=("units_sold", "sum"),
    avg_discount=("discount_pct", "mean"),
    avg_rating=("customer_rating", "mean"),
    avg_delivery_time=("delivery_minutes", "mean"),
    record_count=("units_sold", "size")
)
```
